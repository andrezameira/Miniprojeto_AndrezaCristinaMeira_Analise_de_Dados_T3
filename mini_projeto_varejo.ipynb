### Importando as bibliotecas que serão utilizadas

```python
import pandas as pd
import numpy as np
```

### Criando um DataFrame a paritr do CSV

```python
# Como o separador usado no arquivo não é o padrão que é a vírgula, se fez necessário usar o parâmetro SET
df_varejo = pd.read_csv("base_varejo.csv", sep=";", encoding="utf-8")
```

### Verificando as primeiras linhas do DataFrame

```python
df_varejo.head()
```

### Mantendo apenas as colunas que não estão em branco

```python
df_varejo = df_varejo.dropna(axis=1)
df_varejo
```

### Ajustando o nome das colunas para uma melhor identificá-las

```python
df_varejo.columns = ["Data", "nota_fiscal", "id_cliente", "sexo", "estado_civil", "qtd_filhos", "classe_social", "sku_produto", "categoria_produto", "nome_produto"]
df_varejo
```

### Verificando se há valores ausentes

```python
df_varejo.isna().sum()
```

### Convertendo a coluna Data para o formato Datetime

```python
df_varejo["Data"] = pd.to_datetime(
    df_varejo["Data"],
    dayfirst=True
)

df_varejo
```

### Incluindo uma coluna de ano, para permitir análise anual

```python
df_varejo["Ano_venda"] = df_varejo["Data"].dt.year
df_varejo
```

### Ajustando os valores da coluna estado civil, tirando o número e colocando o status correspondente a cada número

```python
def situacao(num):
    if num == 1:
        return "Casado"
    elif num == 2:
        return "Divorciado"
    elif num == 3:
        return "Separado"
    elif num == 4:
        return "Solteiro"
    elif num == 5:
        return "Viúvo"
    else:
        return "Verificar status"

df_varejo["estado_civil"] = df_varejo["estado_civil"].apply(situacao)
df_varejo
```

### Analisando estatisticamente a coluna de quantidade de filhos

```python
df_varejo["qtd_filhos"].describe()
```

### Calculando a média da quantidade de filhos por classe social

```python
df_varejo.groupby("classe_social")["qtd_filhos"].mean()
```

### Verificando a quantidade de produtos por categoria

```python
df_varejo.groupby("categoria_produto")["nome_produto"].count()
```

### Verificando com estado civil realiza mais compras

```python
df_varejo.groupby("estado_civil")["nota_fiscal"].count()
```

### Verificando qual sexo compra mais

```python
df_varejo.groupby("sexo")["nota_fiscal"].count()
```

### Verificando qual categoria de produto mais vende

```python
df_varejo.groupby("categoria_produto")["nota_fiscal"].count()
```

### Colocando "Não informado" na categoria que não tem nome

```python
df_varejo["categoria_produto"] = df_varejo["categoria_produto"].replace(
    "#N/D",
    "NÃO INFORMADO"
)

df_varejo.groupby("categoria_produto")["nome_produto"].count()
```
